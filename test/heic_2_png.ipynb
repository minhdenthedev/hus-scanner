```python
import os
from PIL import Image
from pillow_heif import register_heif_opener

def convert_heic_to_png(input_folder, output_folder):
    register_heif_opener()

    # os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".heic"):
            heic_file_path = os.path.join(input_folder, filename)
            png_file_path = os.path.join(output_folder, os.path.splitext(filename)[0] + "_raw.png")
            
            try:
                # Mở tệp HEIC và chuyển sang PNG
                with Image.open(heic_file_path) as img:
                    img.save(png_file_path, "PNG")
                print(f"Đã chuyển đổi: {heic_file_path} -> {png_file_path}")
            except Exception as e:
                print(f"Lỗi khi xử lý tệp {heic_file_path}: {e}")

input_folder = "../test_images/unfiltered_pngs/not_named"
output_folder = "../test_images/unfiltered_pngs"

convert_heic_to_png(input_folder, output_folder)
```
